```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
def preprocess_text(text):
  #lowercase
  text = text.lower()
  #remove the puncutation
  text = re.sub(r'[^a-zA-Z\s]','',text)

  words = text.split()
  words =[word for word in words if word not in stop_words]

  #lemmatize
  words =[lemmatizer.lemmatize(word) for word in words]


  return " ".join(words)


```

```python
df['tags']= df['tags'].apply(preprocess_text)
```

```python
df['tags'][1]
```

```python
#covert into vectors
```

```python
#reseting the index

df = df.reset_index(drop = True)
```

```python
# wanting of the indices
indices =pd.Series(df.index,index=df['title']).drop_duplicates()
indices


```

```python
#apply vectorization
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf =TfidfVectorizer(max_features=50000,ngram_range=(1,2),stop_words='english')
```

```python
tfidf_matrix = tfidf.fit_transform(df['tags'])
```

```python
tfidf_matrix.shape
```

```python
tfidf_matrix
```

```python
#import cosine similarity

from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(title,n =10):
  if title not in indices:
    return['Movie not found']

  idx=indices[title]
  sim_score = cosine_similarity(tfidf_matrix[idx],tfidf_matrix).flatten()
  similar_idx = sim_score.argsort()[::-1][1: n+1]

  # Ensure indices are within bounds before passing to iloc
  valid_indices = similar_idx[(similar_idx >= 0) & (similar_idx < len(df))]

  return df['title'].iloc[valid_indices]
```

```python
recommend('Betrayal')
```

```python
recommend('Black Bread')
```

```python
recommend('Black Bread',3)
```

```python
import pickle

pickle.dump(tfidf_matrix,open('tfidf_matrix.pkl','wb'))

pickle.dump(indices,open('indices.pkl','wb'))

df.to_pickle('df.pkl')

pickle.dump(tfidf,open('tfidf.pkl','wb'))
```
